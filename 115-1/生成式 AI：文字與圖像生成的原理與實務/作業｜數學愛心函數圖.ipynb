## 重點說明

把三角函數組合成一條**心形參數曲線**，再用 Matplotlib 做填色、描邊與文字排版。

- **NumPy**：用 `np.linspace()` 產生連續參數 $t$，並計算每個座標點。
- **Matplotlib**：繪製、填滿心形曲線，並加入標題與文字。
- **創意重點**：把數學函數變成一張「I LOVE YOU」視覺圖，讓函數圖形不只是座標線。

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 數學原理

我使用下列參數方程式產生心形：

$$x(t)=16\sin^3(t)$$

$$y(t)=13\cos(t)-5\cos(2t)-2\cos(3t)-\cos(4t)$$

其中 $0\le t\le 2\pi$。當 $t$ 從 0 走到 $2\pi$ 時，$(x(t),y(t))$ 會繞出完整的愛心輪廓。

```python
# 1. 建立參數 t，計算心形曲線的 x、y 座標
t = np.linspace(0, 2 * np.pi, 1200)
x = 16 * np.sin(t)**3
y = 13 * np.cos(t) - 5 * np.cos(2*t) - 2 * np.cos(3*t) - np.cos(4*t)

# 2. 畫出心形：做出外框，再填入顏色
plt.figure(figsize=(8, 8))
plt.plot(x, y, linewidth=12, alpha=0.08, color='#c9184a')
plt.plot(x, y, linewidth=7, alpha=0.14, color='#c9184a')
plt.fill(x, y, color='#ff8fab', alpha=0.92)
plt.plot(x, y, color='#c9184a', linewidth=2.5)

# 3. 在曲線上加一些取樣點，讓函數由很多座標點組成
sample = np.arange(0, len(t), 75)
plt.scatter(x[sample], y[sample], s=18, color='white', alpha=0.8)

# 4. 加入我愛你ㄉ文字
plt.text(0, 1.3, 'I LOVE YOU', ha='center', va='center',
         fontsize=24, weight='bold', color='white')
plt.title('Mathematical Heart Curve', fontsize=16, pad=16)

plt.axis('equal')
plt.axis('off')
plt.show()
```
